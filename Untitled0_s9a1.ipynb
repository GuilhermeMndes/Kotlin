```python
# 🎨 Equipando nosso laboratório de crítica de arte digital

import torch
import torch.nn as nn
import torchvision.transforms as transforms
import torchvision.models as models
from torchvision.datasets import CIFAR10
import numpy as np
import matplotlib.pyplot as plt
from scipy import linalg
from PIL import Image
import warnings

warnings.filterwarnings('ignore')

print("🎨 Crítico de Arte Digital inicializado!")
print("🔬 Laboratório equipado com:")
print("   🧠 Inception-V3 (nosso 'olho treinado')")
print("   📊 Calculadora FID")
print("   ⭐ Medidor Inception Score")
print("   📈 Analisador de qualidade")


# ============================================================
# PARTE 2: PREPARANDO O "OLHO TREINADO" (Inception-V3)
# ============================================================

print("\n👁️ PREPARANDO O 'OLHO TREINADO' (Inception-V3)")
print("=" * 50)

print("🤔 Por que Inception-V3?")
print("   • Treinado em milhões de imagens reais")
print("   • Reconhece 1000 categorias diferentes")
print("   • É como um crítico que viu toda a história da arte")
print("   • Suas 'opiniões' são estatisticamente confiáveis")
print()

# Definindo dispositivo
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

# Carregando Inception-V3 pré-treinado
inception_model = models.inception_v3(
    pretrained=True,
    transform_input=False
)

# Modo avaliação
inception_model.eval()

# Removendo última camada classificadora
inception_model.fc = nn.Identity()

# Enviando modelo para GPU/CPU
inception_model = inception_model.to(device)

print(f"✅ Inception-V3 carregado no: {device}")
print("🎯 Configurado para extrair features")
print("📊 Pronto para analisar qualidade de imagens")


# ============================================================
# PRÉ-PROCESSAMENTO PARA O INCEPTION
# ============================================================

inception_transform = transforms.Compose([
    transforms.Resize((299, 299)),  # Tamanho exigido pelo Inception
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

print("\n🔧 Pré-processamento configurado:")
print("   📐 Redimensionamento: 299x299")
print("   🎨 Normalização: padrão ImageNet")
print("   ✅ Pronto para receber imagens")


# ============================================================
# PARTE 3: EXTRAINDO FEATURES
# ============================================================

def extrair_features_inception(images, model, batch_size=32):
    """
    Extrai features usando Inception-V3
    (como um crítico analisando obras de arte)
    """

    print(f"🔍 Analisando {len(images)} imagens...")

    model.eval()
    features_list = []

    # Processamento em lotes
    for i in range(0, len(images), batch_size):

        batch = images[i:i + batch_size]

        with torch.no_grad():

            # Extraindo features
            features = model(batch)

            # Convertendo para numpy
            features_list.append(features.cpu().numpy())

    # Concatenando tudo
    all_features = np.concatenate(features_list, axis=0)

    print("✅ Análise concluída!")
    print(f"📐 Shape das features: {all_features.shape}")

    return all_features


# ============================================================
# CARREGANDO CIFAR-10
# ============================================================

print("\n📚 Carregando galeria de referência (CIFAR-10)...")

dataset = CIFAR10(
    root='./data',
    train=True,
    download=True,
    transform=inception_transform
)

dataloader = torch.utils.data.DataLoader(
    dataset,
    batch_size=32,
    shuffle=False
)

# Pegando algumas amostras
sample_images = []
sample_labels = []

for i, (images, labels) in enumerate(dataloader):

    sample_images.append(images.to(device))
    sample_labels.extend(labels.numpy())

    # Apenas 4 lotes
    if i >= 3:
        break

# Concatenando imagens
sample_batch = torch.cat(sample_images, dim=0)

print(f"🎨 Amostra carregada: {sample_batch.shape[0]} imagens")


# ============================================================
# EXTRAINDO FEATURES
# ============================================================

print("\n🧠 EXTRAINDO 'OPINIÕES' DO INCEPTION")
print("=" * 45)

sample_features = extrair_features_inception(
    sample_batch,
    inception_model
)


# ============================================================
# ESTATÍSTICAS DAS FEATURES
# ============================================================

print("\n📊 ESTATÍSTICAS DAS FEATURES:")
print(f"   📏 Dimensionalidade: {sample_features.shape[1]} features")
print(f"   📈 Média geral: {sample_features.mean():.4f}")
print(f"   📊 Desvio padrão: {sample_features.std():.4f}")
print(f"   📉 Valor mínimo: {sample_features.min():.4f}")
print(f"   📈 Valor máximo: {sample_features.max():.4f}")


# ============================================================
# VISUALIZAÇÃO DAS FEATURES
# ============================================================

plt.figure(figsize=(12, 4))

# Distribuição geral
plt.subplot(1, 3, 1)

plt.hist(
    sample_features.flatten(),
    bins=50,
    alpha=0.7
)

plt.title('📊 Distribuição das Features')
plt.xlabel('Valor')
plt.ylabel('Frequência')


# Média das features
plt.subplot(1, 3, 2)

plt.plot(sample_features.mean(axis=0)[:100])

plt.title('📈 Média das Features')
plt.xlabel('Índice')
plt.ylabel('Valor Médio')


# Desvio padrão
plt.subplot(1, 3, 3)

plt.plot(sample_features.std(axis=0)[:100])

plt.title('📊 Desvio Padrão')
plt.xlabel('Índice')
plt.ylabel('Desvio')


plt.tight_layout()
plt.show()
```
